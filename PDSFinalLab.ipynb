```python
# Problem 01
"""Load the Titanic dataset using Pandas. Display the first five rows and inspect the dataset. Identify the
numerical and categorical attributes that can be used for visualization."""
import pandas as pd
import matplotlib.pyplot as plt
df=pd.read_csv("train (1).csv")
print(df.head(5))
df.info()
df.isnull().sum()
```

```python
# Problem 02
"""Create a line plot to visualize the average fare paid by passengers across different passenger classes.
Include an appropriate title and axis labels."""

average_fare_by_class = df.groupby('Pclass')['Fare'].mean()

plt.plot(average_fare_by_class.index, average_fare_by_class.values, marker='o')
plt.title('Average Fare by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Average Fare')
plt.show()
```

```python
# Problem 03
"""Create a scatter plot to visualize the relationship between passenger age and fare. Add an appropriate
title and axis labels."""


plt.scatter(df['Age'], df['Fare'])
plt.title('Relationship between Age and Fare')
plt.xlabel('Age')
plt.ylabel('Fare')
plt.show()
```

```python
# Problem 04
"""Create a bar chart to compare the number of male and female passengers in the Titanic dataset. Include
an appropriate title and axis labels."""


gender_counts = df['Sex'].value_counts()
plt.bar(gender_counts.index, gender_counts.values)
plt.title('Number of Male and Female Passengers')
plt.xlabel('Gender')
plt.ylabel('Count')
plt.show()
```

```python
# Problem 05
""" Create a histogram to visualize the distribution of passenger ages. Use an appropriate number of bins and add a title and axis labels."""
plt.hist(df['Age'], bins=20, )
plt.title('Distribution of Passenger Ages')
plt.xlabel('Age')
plt.ylabel('Count')
plt.show()
```

```python
# Problem 06
"""Create a pie chart to represent the proportion of passengers who survived and did not survive. Label the 
categories and display their percentages."""
survival_counts = df['Survived'].value_counts()
plt.pie(survival_counts.values, labels=survival_counts.index, autopct='%1.1f%%')
plt.title('Proportion of Passengers who Survived and Did Not Survive')
plt.show()
```

```python
# Problem 07
"""Create a figure containing two subplots:
1. A histogram showing the distribution of passenger ages.
2. A bar chart showing the number of passengers in each passenger class."""

(ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))
ax1.hist(df['Age'], bins=20, edgecolor = 'black')
ax1.set_title('Distribution of Passenger Ages')
ax1.set_xlabel('Age')
ax1.set_ylabel('Count')
ax2.bar(df['Pclass'].value_counts().index, df['Pclass'].value_counts().values, edgecolor = 'black')
ax2.set_title('Number of Passengers in Each Class')
ax2.set_xlabel('Passenger Class')
ax2.set_ylabel('Count')
```

```python
# Problem 08
"""Create a figure containing three subplots to visualize:
1. The number of passengers by gender using a bar chart.
2. The distribution of passenger fares using a histogram.
3. The survival proportion using a pie chart."""

fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(18, 5))

gender_counts = df['Sex'].value_counts()
ax1.bar(gender_counts.index, gender_counts.values)
ax1.set_title('Number of Male and Female Passengers')
ax1.set_xlabel('Gender')
ax1.set_ylabel('Count')

ax2.hist(df['Fare'], bins=20)
ax2.set_title('Distribution of Passenger Fares')
ax2.set_xlabel('Fare')
ax2.set_ylabel('Count')

survival_counts = df['Survived'].value_counts()
ax3.pie(survival_counts.values, labels=survival_counts.index, autopct='%1.1f%%')
ax3.set_title('Proportion of Passengers who Survived and Did Not Survive')
```
