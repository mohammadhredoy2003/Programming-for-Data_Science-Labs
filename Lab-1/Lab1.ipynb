```python
# Input temperature in Celsius
celsius = float(input("Enter temperature in Celsius: "))

# Convert to Fahrenheit and Kelvin
fahrenheit = (celsius * 9 / 5) + 32
kelvin = celsius + 273.15

print("\n--- Temperature Conversions ---")
print(f"Celsius    : {celsius:.2f} °C")
print(f"Fahrenheit : {fahrenheit:.2f} °F")
print(f"Kelvin     : {kelvin:.2f} K")
```

```python

name = input("Enter student name: ")
marks = []
for i in range(1, 6):
    mark = float(input(f"Enter mark for course {i}: "))
    marks.append(mark)

total = sum(marks)
average = total / len(marks)
highest = max(marks)
lowest = min(marks)

passed_courses = sum(1 for mark in marks if mark >= 50)

print(f"Marks Entered    : {marks}")
print(f"Total Marks      : {total}")
print(f"Average Mark     : {average:.2f}")
print(f"Highest Mark     : {highest}")
print(f"Lowest Mark      : {lowest}")
print(f"Passed Courses   : {passed_courses} out of 5")
```
